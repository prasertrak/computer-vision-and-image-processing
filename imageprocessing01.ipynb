```python
import numpy as np
import matplotlib.pyplot as plt

import skimage                  #"scikit-image, an image processing toolkit in Python"
from skimage import data
```

```python
camera = data.camera()
camera
```

```python
plt.imshow(camera)
```

```python
plt.imshow(camera, 'gray')
```

```python
# ปรับความเข้มของสีของ pixel ที่ต้องการ
# ปรับ 10 แถวแรกให้เป็นสีดำ
camera[:10] = 0
# ปรับ 10 คอลัมน์แรกให้เป็นสีดำ
camera[:, :10] = 0
# ปรับ 100 แถว และ 100 คอลัมน์แรก ให้เป็นสีดำ
camera[:100, :100] = 0
plt.imshow(camera, 'gray')
```

```python
# Pixel ใดใน camera ที่มีความเข้ม < 87
# ให้ปรับเป็นสีขาว
camera = data.camera()
mask = camera < 87
camera[mask] = 255
plt.imshow(camera, 'gray')
```

```python
coins = data.coins()
plt.imshow(coins)
```

```python
from skimage import filters

edges = filters.sobel(coins)
plt.imshow(edges, cmap='gray')
```

```python
img = data.camera()
```

```python
img.shape
```

```python
'''
img.ravel() คือ ฟังก์ชันของ NumPy ที่ใช้สำหรับ "คลี่" หรือแปลงรูปภาพจากอาร์เรย์หลายมิติ
(เช่น 2 มิติ หรือ 3 มิติ) ให้กลายเป็นอาร์เรย์ 1 มิติ (เส้นตรงยาวๆ เส้นเดียว)
โดยไม่มีการคัดลอกข้อมูลใหม่ในหน่วยความจำหากไม่จำเป็น (ทำให้ทำงานเร็วมาก)
'''
img.ravel()
```

```python
'''
วาดกราฟฮิสโตแกรม (Histogram) ของรูปภาพ
เพื่อแสดงการแจกแจงความถี่ของค่าความสว่างในแต่ละพิกเซล โดยมีลักษณะเป็น เส้นขอบสีดำ
'''

plt.hist(img.ravel(), bins=256, histtype='step',
         color='black');
```
