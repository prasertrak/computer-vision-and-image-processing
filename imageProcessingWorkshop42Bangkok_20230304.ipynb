```python
def least_difference(a, b, c):
    diff1 = abs(a - b)
    diff2 = abs(b - c)
    diff3 = abs(a - c)
    return max(diff1, diff2, diff3)
```

```python
print(
least_difference(1, 10, 100),
least_difference(1, 10, 10),
least_difference(5, 6, 7)
)
```

```python
def to_smash(total_candies, n_friends=3):
    """Return จำนวนขนมหวานที่เหลือหลังจากแบ่งให้เพื่อนจำนวนเท่าๆ กัน 3 คน
    >>> to_smash(91)
    1
    """
    return (total_candies % n_friends)

# assert to_smash(91, 4) == 1
print(to_smash(42, 4))
```

```python
# แก้ function ให้ print คำว่า candy, candies ถูกต้องตามจำนวนขนมหวานที่เหลือ

def to_smash(total_candies):
    """Return จำนวนขนมหวานที่เหลือหลังจากแบ่งกับเพื่อนๆ 3 คน จำนวนเท่าๆ กัน

    >>> to_smash(91)
    1
    """
    # print("Splitting", total_candies, "candies")

    # แบบ One line if condition
    # print("Splitting", total_candies, "candy" if total_candies == 1 else "candies")

    # if condition แบบปกติ
    if total_candies == 1:
      print("Splitting", total_candies, "candy")
    else:
      print("Splitting", total_candies, "candies")
    return total_candies % 3

to_smash(54)
```

```python
# nums เป็น list ของตัวเลข
def count_negatives(nums):
  # One line if แบบที่ 1
  # return len([num for num in nums if num < 0])
  # len([-1, -2, -3]) == 3

  # One line if แบบที่ 2
  # return sum([num < 0 for num in nums])
  # sum([True True True]) == 3

  # if condition แบบปกติ
  n_negative = 0
  for num in nums:
    if num < 0:
      n_negative += 1
  return n_negative

nums = [-5,4,-6,7,-9,3]
count_negatives(nums)
```

```python
# Check the image matrix data type (could know the bit depth of the image)
print(myImg.dtype)
# Check the height of image
print(myImg.shape[0])
# Check the width of image
print(myImg.shape[1])
# Check the number of channels of the image [เป็นการผสมสี 3 สี]
print(myImg.shape[2])
```

```python
plt.hist(myImg.ravel(),bins = 256, range = [0,256])
plt.show()
```

```python
color = ('b','g','r')
for i,col in enumerate(color):
    histr = cv.calcHist([myImg],[i],None,[256],[0,256])
    plt.plot(histr,color = col)
    plt.xlim([0,256])
plt.show()
```

```python
gray_image = cv.cvtColor(myImg, cv.COLOR_BGR2GRAY)
noise = cv.cvtColor(myImg, cv.COLOR_BGR2GRAY)
cv2_imshow(gray_image)

#เราแปลงสีเป็นขาวดำได้ แต่แปลงขาวดำเป็นสีไม่ได้
#เนื่องจากขาวดำมี 1 channel สีมี 3 channel
#binaly image = ภาพขาวดำ
#gray image = ภาพเทา
```

```python
# Plot the histogram of the gray image. We could observe that the frequency of
# the image hist has decreased ~ 1/3 of the histogram of color image
plt.hist(gray_image.ravel(),bins = 256, range = [0, 256])
plt.show()
```

```python
# gray_image = noisy("gauss", gray_image)

cv.randn(gray_image, 0, 20)
ex = gray_image+noise
add = cv.add(gray_image,noise)
cv2_imshow(add)
cv2_imshow(ex)
```

```python
kernel = np.ones((3,3),np.float32)/9
#kernel = np.array([[-1,0,1],[-2,0,2],[-1,0,1]])
dst = cv.filter2D(add,-1,kernel)
cv2_imshow(dst)
cv2_imshow(add)
plt.subplot(121),plt.imshow(add, cmap='gray'),plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(dst, cmap='gray'),plt.title('Averaging')
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
kernel = np.ones((11,11),np.float32)/121
dst = cv.filter2D(add,-1,kernel)
cv2_imshow(dst)
```

```python
gaussian = cv.GaussianBlur(add,(5,5),0)
median = cv.medianBlur(add,5)
blur2 = cv.bilateralFilter(add,9,75,75)
cv2_imshow(add)
cv2_imshow(gaussian)
cv2_imshow(median)
cv2_imshow(blur2)

plt.subplot(121),plt.imshow(blur2, cmap='gray'),plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(median, cmap='gray'),plt.title('Blurred')
plt.xticks([]), plt.yticks([])
plt.show()
```
