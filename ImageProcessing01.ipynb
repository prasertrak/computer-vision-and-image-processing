```python
import numpy as np
import matplotlib.pyplot as plt

import skimage
from skimage import data
```

```python
camera = data.camera()
camera
```

```python
plt.imshow(camera)
```

```python
plt.imshow(camera, 'gray')
```

```python
# ปรับความเข้มของสีของ pixel ที่ต้องการ
# ปรับ 10 แถวแรกให้เป็นสีดำ
camera[:10] = 0
# ปรับ 10 คอลัมน์แรกให้เป็นสีดำ
camera[:, :10] = 0
# ปรับ 100 แถว และ 100 คอลัมน์แรก ให้เป็นสีดำ
camera[:100, :100] = 0
plt.imshow(camera, 'gray')
```

```python
# Pixel ใดใน camera ที่มีความเข้ม < 87
# ให้ปรับเป็นสีขาว
camera = data.camera()
mask = camera < 87
camera[mask] = 255
plt.imshow(camera, 'gray')
```

```python
coins = data.coins()
plt.imshow(coins)
```

```python
from skimage import filters

edges = filters.sobel(coins)
plt.imshow(edges, cmap='gray')
```
