```python
# Color Object Detection
# [สมมติว่าคุณมีตัวแปร img ที่เป็น BGR จาก OpenCV เรียบร้อยแล้ว]
# img = cv2.imread('/content/sudoku.jpg')

# 1. แปลงระบบสีจาก BGR เป็น HSV (จำเป็นมากสำหรับการตรวจจับสี)
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# =========================================================================
# 🎯 ส่วนที่ 1: ตรวจจับวัตถุหรือตัวเลข "สีแดง" (Red Color Detection)
# =========================================================================
# สีแดงในระบบ HSV ค่อนข้างพิเศษเพราะค่าเนื้อสี (Hue) จะคาบเกี่ยวอยู่ทั้งช่วงเริ่มต้นและช่วงท้ายของสเกล
# เราจึงนิยมสร้างหน้ากาก (Mask) 2 ช่วงแล้วนำมารวมกัน

# ช่วงที่ 1: สีแดงโทนอ่อน/ส้ม
lower_red1 = np.array([0, 50, 50])
upper_red1 = np.array([10, 255, 255])
mask_red1 = cv2.inRange(hsv, lower_red1, upper_red1)

# ช่วงที่ 2: สีแดงโทนเข้ม/ชมพูเข้ม
lower_red2 = np.array([170, 50, 50])
upper_red2 = np.array([180, 255, 255])
mask_red2 = cv2.inRange(hsv, lower_red2, upper_red2)

# รวมหน้ากากสีแดงทั้งสองช่วงเข้าด้วยกัน
mask_red = cv2.bitwise_or(mask_red1, mask_red2)

# นำหน้ากากไปกรองกับภาพต้นฉบับเพื่อดึงเฉพาะส่วนที่เป็นสีแดงออกมาแสดงผล
detected_red = cv2.bitwise_and(img, img, mask=mask_red)
detected_red_rgb = cv2.cvtColor(detected_red, cv2.COLOR_BGR2RGB)

# =========================================================================
# 🎯 ส่วนที่ 2: ตรวจจับวัตถุหรือตัวเลข "สีดำ/เส้นตาราง" (Black Color Detection)
# =========================================================================
# สีดำจะพิจารณาจากค่าความสว่าง (Value) ที่ต่ำมากๆ เป็นหลัก
lower_black = np.array([0, 0, 0])
upper_black = np.array([180, 255, 70]) # ปรับตัวเลข 70 ขึ้น-ลงได้ตามความมืดของภาพ
mask_black = cv2.inRange(hsv, lower_black, upper_black)

detected_black = cv2.bitwise_and(img, img, mask=mask_black)
detected_black_rgb = cv2.cvtColor(detected_black, cv2.COLOR_BGR2RGB)

# =========================================================================
# 🖼️ การแสดงผลเปรียบเทียบด้วย Matplotlib (Subplot)
# =========================================================================
plt.figure(figsize=(15, 5))

# ช่องที่ 1: แสดงภาพต้นฉบับ
plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

# ช่องที่ 2: แสดงผลลัพธ์ที่กรองเฉพาะสีแดง (จะเห็นเฉพาะตัวเลขสีแดง)
plt.subplot(1, 3, 2)
plt.imshow(detected_red_rgb)
plt.title('Detected Red Objects')
plt.axis('off')

# ช่องที่ 3: แสดงผลลัพธ์ที่กรองเฉพาะสีดำ (จะเห็นเฉพาะเส้นตารางและตัวเลขสีดำ)
plt.subplot(1, 3, 3)
plt.imshow(detected_black_rgb)
plt.title('Detected Black Objects')
plt.axis('off')

plt.show()
```

```python
# Scaling_w,H
'''
m = np.float32([0.5,0,0],
                [0,0.5,0])
dst = cv2.warpAffine(rgb, m, (rgb.shape[1],rgb.shape[0]))
plt.subplot(1,2,1);plt.imshow(rgb)
plt.subplot(1,2,2);plt.imshow(rgb)
'''

# 1. แก้ไขตรงนี้: เติมวงเล็บเหลี่ยมครอบให้กลายเป็นเมทริกซ์ 2 มิติที่ถูกต้อง
m = np.float32([[0.5, 0, 0],
                [0, 0.5, 0]])

# 2. ทำการประมวลผลย่อขนาดภาพด้วย warpAffine
# (ภาพย่อลงเหลือ 50% แต่พื้นที่เฟรมภาพขนาดเท่าเดิม)
dst = cv2.warpAffine(rgb, m, (rgb.shape[1], rgb.shape[0]))

# 3. ตั้งขนาดหน้าต่างแสดงผลรวม
plt.figure(figsize=(10, 5))

# ช่องที่ 1: แสดงภาพต้นฉบับ (Original)
plt.subplot(1, 2, 1)
plt.imshow(rgb)
plt.title('Original Image')
plt.axis('off')

# ช่องที่ 2: แก้ไขตรงนี้ เปลี่ยนจาก rgb เป็น dst เพื่อดูภาพที่ย่อขนาดแล้ว (Scaled)
plt.subplot(1, 2, 2)
plt.imshow(dst)
plt.title('Scaled Image (0.5x)')
plt.axis('off')

plt.show()
```

```python
# Translation + Scaling_w,H
# 1. สร้างเมทริกซ์ที่รวมทั้งการย่อขนาด (0.5) และการเลื่อนตำแหน่ง (X:+100, Y:+50)
# อย่าลืมใต้วงเล็บเหลี่ยมครอบให้กลายเป็น 2 มิติ [[ ... ]]
m = np.float32([[0.5,   0, 100],  # [Scale_X,       0, Trans_X]
                [  0, 0.5,  50]]) # [      0, Scale_Y, Trans_Y]

# 2. ประมวลผลภาพด้วย Warp Affine ครั้งเดียวได้ผลลัพธ์ทั้งสองอย่าง
dst = cv2.warpAffine(rgb, m, (rgb.shape[1], rgb.shape[0]))

# 3. แสดงผลเปรียบเทียบภาพดั้งเดิม กับ ภาพที่ย่อและขยับแล้ว
plt.figure(figsize=(10, 5))

# ช่องที่ 1: ภาพต้นฉบับ
plt.subplot(1, 2, 1)
plt.imshow(rgb)
plt.title('Original Image')
plt.axis('off')

# ช่องที่ 2: ภาพที่ทำ Translation + Scaling เรียบร้อยแล้ว
plt.subplot(1, 2, 2)
plt.imshow(dst)
plt.title('Scaled (0.5x) + Translated')
plt.axis('off')

plt.show()
```

```python
# Rotation
# 1. หาพิกัดจุดกึ่งกลางของภาพ (Center) เพื่อใช้เป็นจุดหมุน
(h, w) = rgb.shape[:2]
center = (w // 2, h // 2)

# 2. คำนวณหาเมทริกซ์การหมุน m ด้วยฟังก์ชันของ OpenCV
# โครงสร้าง: cv2.getRotationMatrix2D(จุดหมุน, องศาที่ต้องการหมุน, อัตราการย่อขยายภาพ)
# * หมายเหตุ: ค่าองศาที่เป็นบวก = หมุนทวนเข็มนาฬิกา / ค่าลบ = หมุนตามเข็มนาฬิกา
angle = 45
scale = 1.0  # คงขนาดภาพเท่าเดิม (ถ้าอยากย่อไปพร้อมกันปรับเป็น 0.5 ได้ครับ)
m = cv2.getRotationMatrix2D(center, angle, scale)

# 3. นำเมทริกซ์ m ไปประมวลผลภาพด้วย Warp Affine เหมือนเดิม
dst = cv2.warpAffine(rgb, m, (w, h))

# 4. แสดงผลเปรียบเทียบภาพดั้งเดิม กับ ภาพที่หมุนแล้ว
plt.figure(figsize=(10, 5))

# ช่องที่ 1: ภาพต้นฉบับ
plt.subplot(1, 2, 1)
plt.imshow(rgb)
plt.title('Original Image')
plt.axis('off')

# ช่องที่ 2: ภาพที่ทำ Rotation เรียบร้อยแล้ว
plt.subplot(1, 2, 2)
plt.imshow(dst)
plt.title(f'Rotated Image ({angle}°)')
plt.axis('off')

plt.show()
```

```python
# shear x
# 1. กำหนดค่าความบิดเอนตามแกน X (Shear Factor)
shear_x = 0.2

# 2. สร้างเมทริกซ์การบิด Shear X
# (อย่าลืมวงเล็บเหลี่ยมครอบซ้อน 2 ชั้นเด็ดขาดครับ)
m = np.float32([[1, shear_x, 0],
                [0,    1,    0]])

# 3. ประมวลผลภาพด้วย Warp Affine
(h, w) = rgb.shape[:2]
dst = cv2.warpAffine(rgb, m, (w, h))

# 4. แสดงผลเปรียบเทียบภาพดั้งเดิม กับ ภาพที่ทำ Shear X แล้ว
plt.figure(figsize=(10, 5))

# ช่องที่ 1: ภาพต้นฉบับ
plt.subplot(1, 2, 1)
plt.imshow(rgb)
plt.title('Original Image')
plt.axis('off')

# ช่องที่ 2: ภาพที่ถูกบิดเฉียงในแนวแกน X
plt.subplot(1, 2, 2)
plt.imshow(dst)
plt.title(f'Shear X ({shear_x})')
plt.axis('off')

plt.show()

```
